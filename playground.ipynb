# SIITec Hackathon 2569 · ลองเล่น OWLv2
**AI ที่หาของในภาพจากคำที่เราพิมพ์**

**VLM (Vision-Language Model)** คือ AI ที่เข้าใจทั้ง *ภาพ* และ *ข้อความ* ในตัวเดียว
**OWLv2** (จาก Google) เป็น VLM ที่หาของในภาพได้: มันแบ่งภาพเป็นช่องเล็กๆ หลายพันช่อง แล้วเทียบแต่ละช่องกับคำที่เราพิมพ์
ช่องไหนเข้ากับคำได้ดี ก็ได้คะแนนสูง แล้วถูกวาดกรอบ — **ไม่ต้องสอน AI ใหม่ แค่เปลี่ยนคำ**

ลองดู
- `a dog` → `a black dog` → `a dog's tongue` · ยิ่งเจาะจง กรอบเปลี่ยนไหม?
- `a person` → `a man in a red shirt` → `a child`
- เลื่อน **ความมั่นใจขั้นต่ำ** ขึ้นลง · ต่ำ = เจอเยอะแต่ผิดเยอะ, สูง = เจอน้อยแต่แม่น
- พิมพ์ภาษาไทยดู แล้วเทียบกับภาษาอังกฤษ
- อัปโหลดรูปของเราเอง
